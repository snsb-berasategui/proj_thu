```r
library(ggplot2)
library(dplyr)

dat <- read.csv("data/try_reduced.csv")
head(dat)
```

How we will create a figure:

```r
dat |> filter(SpeciesName == "Fagus sylvatica",
              TraitName == "Seed dry mass") |>
  ggplot(aes(StdValue)) + geom_histogram(fill = "pink") +
  scale_x_continuous(trans = "log10") +
  theme_light()
```
